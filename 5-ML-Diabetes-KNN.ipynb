### ML Practical 5 (B-5): Diabetes Prediction using K-Nearest Neighbors (KNN)
**Problem Statement:**
Implement K-Nearest Neighbors algorithm on `diabetes.csv` dataset. Compute confusion matrix, accuracy, error rate, precision, and recall.

```python
# Cell 1: Import Libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, classification_report
```

```python
# Cell 2: Load & Inspect Dataset
df = pd.read_csv("diabetes.csv")
df.head()
```

```python
# Cell 3: Statistical Summary
df.describe()
```

```python
# Cell 4: Separate Features (X) and Target (Y)
X = df.drop("Outcome", axis=1)
Y = df["Outcome"]
```

```python
# Cell 5: Train-Test Split (80% Train, 20% Test)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, random_state=42)
```

```python
# Cell 6: Feature Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Cell 7: Train KNN Model & Predict
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, Y_train)
Y_pred = knn.predict(X_test)
```

```python
# Cell 8: Compute Confusion Matrix, Accuracy, Error Rate, Precision & Recall
cm = confusion_matrix(Y_test, Y_pred)
print("Confusion Matrix:\n", cm)

accuracy = accuracy_score(Y_test, Y_pred)
error_rate = 1 - accuracy
precision = precision_score(Y_test, Y_pred)
recall = recall_score(Y_test, Y_pred)

print("\nAccuracy   :", accuracy)
print("Error Rate :", error_rate)
print("Precision  :", precision)
print("Recall     :", recall)

cr = classification_report(Y_test, Y_pred)
print("\nClassification Report:\n", cr)
```
