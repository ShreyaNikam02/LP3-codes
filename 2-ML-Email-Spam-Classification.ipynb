### ML Practical 2 (B-II): Email Spam Classification using KNN and SVM
**Problem Statement:**
Classify the email using binary classification (`0`: Not Spam, `1`: Spam) using **K-Nearest Neighbors (KNN)** and **Support Vector Machine (SVM)**, and analyze their performance.

```python
# Cell 1: Import Libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

```python
# Cell 2: Load & Inspect Dataset
df = pd.read_csv("emails.csv")
df.head()
```

```python
# Cell 3: Statistical Summary
df.describe()
```

```python
# Cell 4: Separate Features (X) and Target (Y)
X = df.drop(columns=['Email No.', 'Prediction'])
Y = df['Prediction']
X.head()
```

```python
# Cell 5: Train-Test Split (80% Train, 20% Test)
Xtrain, Xtest, Ytrain, Ytest = train_test_split(X, Y, test_size=0.2, random_state=1)
```

```python
# Cell 6: Train K-Nearest Neighbors (KNN) Model
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(Xtrain, Ytrain)
knn_predictions = knn_model.predict(Xtest)
```

```python
# Cell 7: Evaluate KNN Performance
knn_accuracy = accuracy_score(Ytest, knn_predictions)
print("KNN Accuracy:", knn_accuracy)

knn_conf_matrix = confusion_matrix(Ytest, knn_predictions)
print("\nKNN Confusion Matrix:\n", knn_conf_matrix)

knn_classification_report = classification_report(Ytest, knn_predictions)
print("\nKNN Classification Report:\n", knn_classification_report)
```

```python
# Cell 8: Train Support Vector Machine (SVM) Model
svm_model = SVC(kernel="linear")
svm_model.fit(Xtrain, Ytrain)
svm_predictions = svm_model.predict(Xtest)
```

```python
# Cell 9: Evaluate SVM Performance
svm_accuracy = accuracy_score(Ytest, svm_predictions)
print("SVM Accuracy:", svm_accuracy)

svm_confusion_matrix = confusion_matrix(Ytest, svm_predictions)
print("\nSVM Confusion Matrix:\n", svm_confusion_matrix)

svm_classification_report = classification_report(Ytest, svm_predictions)
print("\nSVM Classification Report:\n", svm_classification_report)
```
