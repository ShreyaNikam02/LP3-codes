### ML Practical 6 (B-6): Sales Data Clustering (K-Means & Hierarchical Clustering)
**Problem Statement:**
Implement K-Means clustering / hierarchical clustering on `sales_data_sample.csv` dataset. Determine the number of clusters using the elbow method.

```python
# Cell 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
```

```python
# Cell 2: Load & Inspect Dataset
df = pd.read_csv('sales_data_sample.csv', encoding='latin1')
df.head()
```

```python
# Cell 3: Dataset Info & Summary
df.info()
df.describe()
```

```python
# Cell 4: Select Numerical Features
features = ['QUANTITYORDERED', 'SALES']
X = df[features]
X.isnull().sum()
```

```python
# Cell 5: Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Cell 6: Elbow Method to Determine Optimal Number of Clusters
wcss = []
for i in range(1, 11):
    model = KMeans(n_clusters=i, random_state=42)
    model.fit(X_scaled)
    wcss.append(model.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

```python
# Cell 7: K-Means Clustering (Optimal k = 3)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)

df['KMeans_Cluster'] = kmeans_labels
print(df[['KMeans_Cluster']].head())
```

```python
# Cell 8: Plot K-Means Clusters
plt.scatter(df['QUANTITYORDERED'], df['SALES'], c=kmeans_labels, cmap='viridis')
plt.xlabel('Quantity Ordered')
plt.ylabel('Sales')
plt.title('K-Means Clustering')
plt.colorbar(label='Cluster')
plt.show()
```

```python
# Cell 9: Hierarchical Clustering (n_clusters = 3)
hc = AgglomerativeClustering(n_clusters=3)
hc_labels = hc.fit_predict(X_scaled)

df['Hierarchical_Cluster'] = hc_labels
print(df[['Hierarchical_Cluster']].head())
```

```python
# Cell 10: Plot Hierarchical Clusters
plt.scatter(df['QUANTITYORDERED'], df['SALES'], c=hc_labels, cmap='viridis')
plt.xlabel('Quantity Ordered')
plt.ylabel('Sales')
plt.title('Hierarchical Clustering')
plt.colorbar(label='Cluster')
plt.show()
```

```python
# Cell 11: Evaluate K-Means & Hierarchical Clustering
print("K-Means Evaluation")
print("Silhouette Score:", silhouette_score(X_scaled, kmeans_labels))
print("Calinski-Harabasz Score:", calinski_harabasz_score(X_scaled, kmeans_labels))
print("Davies-Bouldin Index:", davies_bouldin_score(X_scaled, kmeans_labels))

print("\nHierarchical Clustering Evaluation")
print("Silhouette Score:", silhouette_score(X_scaled, hc_labels))
print("Calinski-Harabasz Score:", calinski_harabasz_score(X_scaled, hc_labels))
print("Davies-Bouldin Index:", davies_bouldin_score(X_scaled, hc_labels))
```
